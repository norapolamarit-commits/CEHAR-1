```python
import json  # TH: นำเข้าเครื่องมืออ่านและเขียน JSON | EN: Import JSON reading and writing tools.
import os  # TH: นำเข้าเครื่องมือระบบปฏิบัติการ | EN: Import operating-system utilities.
import random  # TH: นำเข้าเครื่องมือสุ่มของ Python | EN: Import Python random utilities.
from pathlib import Path  # TH: นำเข้าคลาสจัดการพาธ | EN: Import the path-management class.
import numpy as np  # TH: นำเข้า NumPy สำหรับข้อมูลตัวเลข | EN: Import NumPy for numerical data.
from google.colab import drive  # TH: นำเข้าเครื่องมือเชื่อม Google Drive | EN: Import the Google Drive connector.
drive.mount("/content/drive")  # TH: เชื่อม Google Drive เข้ากับ Colab | EN: Mount Google Drive in Colab.
PROJECT_ROOT = Path("/content/drive/MyDrive/CEHARPS")  # TH: กำหนดโฟลเดอร์หลักของโครงการ | EN: Define the project root folder.
PROJECT_ROOT.mkdir(parents=True, exist_ok=True)  # TH: สร้างโฟลเดอร์หลักหากยังไม่มี | EN: Create the root folder when absent.
```

```python
CONFIG = {  # TH: เริ่มกำหนดค่ากลางของทุกสมุดงาน | EN: Start the shared configuration for all notebooks.
    "project_root": str(PROJECT_ROOT),  # TH: บันทึกพาธหลักเป็นข้อความ | EN: Store the root path as text.
    "seed": 42,  # TH: กำหนดค่าเมล็ดสุ่มเพื่อทำซ้ำผล | EN: Set the random seed for reproducibility.
    "health_mode": "demo",  # TH: ใช้ demo เพื่อทดสอบ หรือ real สำหรับข้อมูลจริง | EN: Use demo for testing or real for real data.
    "health_target": "MHI",  # TH: กำหนดตัวแปรเป้าหมายสุขภาพ | EN: Define the health target variable.
    "health_real_csv": str(PROJECT_ROOT / "data/health/health_features_real.csv"),  # TH: ระบุไฟล์สุขภาพจริง | EN: Point to the real health-data file.
    "segmentation_epochs": 5,  # TH: กำหนดจำนวนรอบฝึก segmentation | EN: Set the segmentation training epochs.
    "batch_size": 4,  # TH: กำหนดจำนวนภาพต่อหนึ่งแบตช์ | EN: Set the number of images per batch.
    "image_size": 128,  # TH: กำหนดขนาดภาพเข้าโมเดล | EN: Set the model input image size.
    "num_workers": 0,  # TH: ปิด worker เพิ่มเพื่อลดปัญหากับ Drive | EN: Disable extra workers to avoid Drive issues.
    "learning_rate": 0.0003,  # TH: กำหนดอัตราการเรียนรู้ | EN: Set the learning rate.
    "early_stopping_patience": 3,  # TH: กำหนดจำนวนรอบรอก่อนหยุดเร็ว | EN: Set early-stopping patience.
    "encoder_name": "resnet34",  # TH: กำหนด backbone ของโมเดลภาพ | EN: Set the image-model backbone.
    "encoder_weights": "imagenet",  # TH: ใช้น้ำหนักเริ่มต้นจาก ImageNet | EN: Use ImageNet initialization weights.
    "xgb_device": "cpu",  # TH: ใช้ CPU เพื่อความเข้ากันได้; เปลี่ยนเป็น cuda ได้ | EN: Use CPU for compatibility; cuda is optional.
    "bootstrap_models": 10,  # TH: กำหนดจำนวนโมเดลสำหรับช่วงความเชื่อมั่น | EN: Set the number of uncertainty models.
    "restoration_budget": 1000000.0,  # TH: กำหนดงบประมาณตัวอย่าง | EN: Set the example restoration budget.
    "pathumma_model_id": "nectec/pathumma-thaillm-8b-think-3.0.0",  # TH: กำหนด Pathumma รุ่นหลักสำหรับ Chatbot | EN: Set the primary Pathumma model for the chatbot.
    "rag_embedding_model_id": "BAAI/bge-m3",  # TH: กำหนดโมเดล embedding ภาษาไทยหลายภาษา | EN: Set the multilingual Thai-capable embedding model.
    "rag_chunk_chars": 1000,  # TH: กำหนดจำนวนอักขระต่อส่วนเอกสาร | EN: Set characters per document chunk.
    "rag_chunk_overlap": 150,  # TH: กำหนดส่วนซ้อนทับระหว่างชิ้นเอกสาร | EN: Set overlap between document chunks.
    "rag_top_k": 5,  # TH: กำหนดจำนวนชิ้นข้อมูลที่ค้นคืน | EN: Set the number of retrieved chunks.
    "rag_min_score": 0.35,  # TH: กำหนดคะแนนขั้นต่ำก่อนอนุญาตให้ตอบ | EN: Set the minimum retrieval score required to answer.
    "rag_max_new_tokens": 512,  # TH: จำกัดความยาวคำตอบของ Chatbot | EN: Limit chatbot response length.
}  # TH: ปิดพจนานุกรมค่ากลาง | EN: Close the shared configuration dictionary.
CONFIG_PATH = PROJECT_ROOT / "config.json"  # TH: กำหนดตำแหน่งไฟล์ตั้งค่า | EN: Define the configuration-file path.
CONFIG_PATH.write_text(json.dumps(CONFIG, ensure_ascii=False, indent=2), encoding="utf-8")  # TH: เขียนค่ากลางลง Drive | EN: Write the configuration to Drive.
for folder in ["data/raw", "data/health", "data/segmentation", "data/processed", "artifacts", "reports"]:  # TH: วนสร้างโฟลเดอร์ที่จำเป็น | EN: Iterate through required folders.
    (PROJECT_ROOT / folder).mkdir(parents=True, exist_ok=True)  # TH: สร้างโฟลเดอร์ปัจจุบัน | EN: Create the current folder.
print(json.dumps(CONFIG, ensure_ascii=False, indent=2))  # TH: แสดงค่าที่บันทึกเพื่อตรวจสอบ | EN: Display the saved configuration for review.
```

```python
SEED = int(CONFIG["seed"])  # TH: อ่านค่าเมล็ดสุ่มเป็นจำนวนเต็ม | EN: Read the random seed as an integer.
random.seed(SEED)  # TH: ตั้งค่าเมล็ดสุ่มของ Python | EN: Seed Python's random generator.
np.random.seed(SEED)  # TH: ตั้งค่าเมล็ดสุ่มของ NumPy | EN: Seed NumPy's random generator.
os.environ["PYTHONHASHSEED"] = str(SEED)  # TH: ควบคุมการสุ่มของค่าแฮช | EN: Control hash randomization.
assert CONFIG_PATH.exists()  # TH: ยืนยันว่าไฟล์ตั้งค่าถูกสร้างแล้ว | EN: Confirm that the configuration file exists.
print("พร้อมใช้งาน / Ready:", CONFIG_PATH)  # TH: แจ้งว่าการตั้งค่าเสร็จสมบูรณ์ | EN: Report that setup is complete.
```
